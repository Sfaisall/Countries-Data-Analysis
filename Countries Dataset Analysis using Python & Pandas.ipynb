# Analysis on Population

### Name of country and their capital which has highest population ?

```python
print(df[df["population"] == df["population"].max()]["country"])
print(df[df["population"] == df["population"].max()]["capital_city"]) 
```

### Name of country and their capital which has Lowest population ?

```python
print(df[df["population"] == df["population"].min()]["country"]) 
print(df[df["population"] == df["population"].min()]["capital_city"])
```

### Top 10 most populated country in world.

```python
most_populated = df.nlargest(10,"population")[["country","population"]]
most_populated
```

### Top 10 least Populated Country in the World.

```python
df.nsmallest(10,"population")[["country","population"]]
```

### Average Population of Country (in Crores upto 2 decimal)

```python
avg_population_crore = df["population"].mean()/10000000
print(f"Average Population of a country is {avg_population_crore:.2f} Crore")
```

### Which country has maximum Agricultural land

```python
df[df["agricultural_land"] == df['agricultural_land'].max()][["country","agricultural_land"]]
```

### Which country has Maximum female Population

```python
df[df["population_female"] == df["population_female"].max()][["country","population_female"]]
```

### How many countries have "republic" in their country name(By using User_defined function)

```python
count = 0
def check_republic(txt):
    global count
    if  "republic" in txt.lower():
        count += 1
    return txt

df["country_long"].apply(check_republic)
print(count)
```

### How many countries have "republic" in their country name

```python
print(df["country_long"].str.contains("Republic", case=False).sum()) # it output in boolean 
# true -->1
#false --->0
```
